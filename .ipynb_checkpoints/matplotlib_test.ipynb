#universal categorical


```python
import pandas as pd 
data = {
    "salary": [ 250000, 300000, 370000, 340000, 450000, 550000, 340000, 56000]
}

df  = pd.DataFrame(data)

df["dept"] = ['HR', 'IT', 'FINANCE', 'HR', 'FINANCE', 'IT', 'HR', 'FINANCE']

df.head()

df.info()
```

```python
#pie chart

count  = df["dept"].value_counts()
count
plt.pie(count)
plt.show()
```

```python
plt.pie(count, labels = count.index, autopct= "1.1f")

plt.show()
```
